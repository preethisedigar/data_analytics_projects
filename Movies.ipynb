**Subtask 3.3: Find IMDb Top 250**

Create a new dataframe `IMDb_Top_250` and store the top 250 movies with the highest IMDb Rating (corresponding to the column: `imdb_score`). Also make sure that for all of these movies, the `num_voted_users` is greater than 25,000.

Also add a `Rank` column containing the values 1 to 250 indicating the ranks of the corresponding films.

```python
# Write your code for extracting the top 250 movies as per the IMDb score here. Make sure that you store it in a new dataframe
# and name that dataframe as 'IMDb_Top_250'
filtered_df = df[df['num_voted_users'] > 25000]
sorted_df = filtered_df.sort_values(by='imdb_score', ascending=False)
top_250_df = sorted_df.head(250)
top_250_df
```

#### <font color='red'>Question 7: Suppose movies are divided into 5 buckets based on the IMDb ratings: </font>

<font color = 'red'> Which bucket holds the maximum number of movies from *IMDb_Top_250*? </font>

```python
top_250_df = pd.DataFrame(top_250_df)
top_250_df
bins = [7.5, 8, 8.5, 9, 9.5, 10]
labels = ['7.5-8', '8-8.5', '8.5-9', '9-9.5', '9.5-10']
```

```python
top_250_df['Rating_Bucket'] = pd.cut(top_250_df['imdb_score'], bins=bins, labels=labels, right=False)
bucket_counts = top_250_df['Rating_Bucket'].value_counts()
max_bucket = bucket_counts.idxmax()
max_count = bucket_counts.max()
print(max_bucket )
print(max_count)
```

**Subtask 3.4: Find the critic-favorite and audience-favorite actors**

   1. Create three new dataframes namely, `Meryl_Streep`, `Leo_Caprio`, and `Brad_Pitt` which contain the movies in which the actors: 'Meryl Streep', 'Leonardo DiCaprio', and 'Brad Pitt' are the lead actors. Use only the `actor_1_name` column for extraction. Also, make sure that you use the names 'Meryl Streep', 'Leonardo DiCaprio', and 'Brad Pitt' for the said extraction.
   2. Append the rows of all these dataframes and store them in a new dataframe named `Combined`.
   3. Group the combined dataframe using the `actor_1_name` column.
   4. Find the mean of the `num_critic_for_reviews` and `num_user_for_review` and identify the actors which have the highest mean.

```python
# Write your code for creating three new dataframes here
Meryl_Streep =  df[df['actor_1_name'] == 'Meryl Streep'] # Include all movies in which Meryl_Streep is the lead
Meryl_Streep.shape[0]
```

```python
Leo_Caprio = df[df['actor_1_name'] == 'Leonardo DiCaprio']
Leo_Caprio.shape[0]    # Include all movies in which Leo_Caprio is the leadLeo_Caprio
```

```python
Brad_Pitt = Brad_Pitt = df[df['actor_1_name'] == 'Brad Pitt']
Brad_Pitt.shape[0]  # Include all movies in which Brad_Pitt is the lead
```

```python
# Write your code for combining the three dataframes here
Combined = pd.concat([Meryl_Streep, Leo_Caprio, Brad_Pitt])
Combined
```

```python
# Write your code for grouping the combined dataframe here
grouped = Combined.groupby('actor_1_name')[['num_critic_for_reviews', 'num_user_for_reviews']].mean()
grouped
```

```python
# Write the code for finding the mean of critic reviews and audience reviews here
highest_mean_criti = grouped['num_critic_for_reviews'].idxmax()
highest_mean_user = grouped['num_user_for_reviews'].idxmax()

```
