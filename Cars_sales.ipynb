```python
sales.Sales = sales.Sales.floordiv(1000)
sales.head()
```

```python
sales.rename(columns = {'Sales':"Sales in Thousand"}, inplace = True)
sales.head()
```

```python
total_sum = sales.Profit.sum()
total_sum
```

```python
sales['Profit %'] = sales.Profit.apply(lambda x:x/total_sum * 100)
```

```python
sales.head()
```

```python
# to make the negative Profit % as NAN
import numpy as np
sales.loc[sales['Profit %'] < 0 ,['Profit %']] = np.nan
sales.head()
```

```python
sales.loc[sales['Profit'] < 0 ,['Profit']] = np.nan
sales.head()
```
