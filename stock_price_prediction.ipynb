```python
#import necessary libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
#read the file
data=pd.read_csv('stock_price.csv',encoding='latin1' )
```

IN the given Dataset file the Description and Customer id columns has null values(missing values).

As we know customer id is unique and cannot be reasonably imputed , might lead to ambiguity.so its better to drop missing values rows,its meaning less to find mean or median for id column.

Description missing values is not appropriate to fill unknown or nan so dropping missing values rows.


```python
data=data.dropna()
data.shape
```

```python
data.isnull().sum()
```

```python
#check duplicates
dup_rows=data[data.duplicated()]
dup_rows
```

```python
data=data.drop_duplicates()
data.shape
```

```python
dup_rows=data[data.duplicated()]
```

```python
data.head()
```

```python
data.describe()
```

```python
data.dtypes
```

```python
data.columns
```

```python
# Create a new column TotalAmount
data['TotalAmount'] = data['Quantity'] * data['UnitPrice']
data.head()
```

```python
# Remove rows where Quantity or UnitPrice is negative
data = data[(data['Quantity'] > 0) & (data['UnitPrice'] > 0)]
```

```python
data.shape
```

```python
data['InvoiceDate'] = pd.to_datetime(data['InvoiceDate'],format='%d-%m-%Y %H:%M')
```

```python
# Extract year, month, day, and weekday
data['Year'] = data['InvoiceDate'].dt.year
data['Month'] = data['InvoiceDate'].dt.month
data['Day'] = data['InvoiceDate'].dt.day
```

```python
data.head()
```

```python
data['CustomerID'].nunique()
```

```python
# Standardize relevant features
scaler = StandardScaler()
data[['Quantity', 'UnitPrice', 'TotalAmount']] = scaler.fit_transform(data[['Quantity', 'UnitPrice', 'TotalAmount']])
```

Vizualize the data

```python
# Visualize the distribution of TotalAmount
plt.figure(figsize=(10, 6))
sns.histplot(data['TotalAmount'], kde=True)
plt.title('Distribution of Total Amount')
plt.xlabel('Total Amount')
plt.ylabel('Frequency')
plt.show()
```
